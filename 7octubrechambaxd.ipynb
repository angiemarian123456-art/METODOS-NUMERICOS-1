```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
  return 3*x**2-np.exp(x)
```

```python
xx=np.linspace(-2,2)
plt.plot(xx,f(xx),color='red')
plt.axhline(y=0,color='blue')
plt.axvline(x=0,color='blue')
```

```python
def g(x):
  return np.sqrt(np.exp(x)/3)
```

```python
g(0)
```

```python
def p_fijo(p0,i):
    p0=0.5

    lista=[]
    error=[]

    for i in range (i):
        p1=g(p0)
        lista.append(p1)
        e=np.abs(p1-p0)
        error.append(e)
        p0=p1
    return lista,error
```

```python
xx=np.linspace(-2,2)
plt.plot(xx,g(xx),color='red')
plt.axhline(y=0,color='black')
plt.axvline(x=0,color='black')
```

```python
p=p_fijo(0.5,30)
```
